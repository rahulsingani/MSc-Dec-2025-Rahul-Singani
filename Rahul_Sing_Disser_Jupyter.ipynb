```python
# ============================================
# Import Required Libraries
# ============================================

# Data manipulation
import pandas as pd
import numpy as np

# Data visualization
import matplotlib.pyplot as plt
import seaborn as sns

# Date and time handling
from datetime import datetime

# Display plots within the notebook
%matplotlib inline

# Optional: Improve the appearance of plots
sns.set_theme(style="whitegrid")

# Ignore warning messages
import warnings
warnings.filterwarnings('ignore')
```

```python
# ============================================
# Load the Dataset
# ============================================

# Read the Excel dataset
df = pd.read_excel("E-Commerce DataSet.xlsx")

# Display the first five records
print("First Five Records:")
display(df.head())

# Check the dataset dimensions
print("Dataset Dimensions (Rows, Columns):", df.shape)
```

```python
# ============================================
# Initial Data Exploration
# ============================================

# Display dataset information
print("Dataset Information:")
df.info()

# Display data types
print("\nData Types:")
print(df.dtypes)

# View summary statistics for numerical columns
print("\nSummary Statistics (Numerical Columns):")
display(df.describe())

# View summary statistics for all columns (including categorical)
print("\nSummary Statistics (All Columns):")
display(df.describe(include='all'))

# Examine column names
print("\nColumn Names:")
print(df.columns.tolist())
```

```python
# ============================================
# Check Data Quality
# ============================================

# Identify missing values
print("Missing Values:")
display(df.isnull().sum())

# Calculate the percentage of missing values
print("\nPercentage of Missing Values:")
display((df.isnull().sum() / len(df) * 100).round(2))

# Check duplicate records
duplicate_count = df.duplicated().sum()

print("\nNumber of Duplicate Records:")
print(duplicate_count)

# Count unique values in each column
print("\nUnique Values in Each Column:")
display(df.nunique())

# Inspect inconsistent entries

# Check for product quantities less than or equal to zero
print("\nRecords with Product Quantity <= 0:")
display(df[df['prod_qty'] <= 0])

# Check for missing product SKUs
print("\nRecords with Missing Product SKU:")
display(df[df['prod_sku'].isnull()])

# Check for missing shipment dates
print("\nRecords with Missing Shipment Date:")
display(df[df['shipped_at'].isnull()].head())

# Check for order dates occurring after shipment dates
print("\nRecords where Order Date is After Shipment Date:")
display(df[df['order_date'] > df['shipped_at']])
```

```python
# ============================================
# Data Cleaning
# ============================================

# Create a copy of the original dataset
df_clean = df.copy()

# ----------------------------------
# Remove duplicate records
# ----------------------------------
df_clean.drop_duplicates(inplace=True)

# ----------------------------------
# Handle missing values
# ----------------------------------

# Fill missing shipped_at values with the median date
median_shipped_date = df_clean['shipped_at'].median()
df_clean['shipped_at'] = df_clean['shipped_at'].fillna(median_shipped_date)

# Fill missing prod_sku values with the mode
mode_sku = df_clean['prod_sku'].mode()[0]
df_clean['prod_sku'] = df_clean['prod_sku'].fillna(mode_sku)

# ----------------------------------
# Correct data types
# ----------------------------------
df_clean['order_date'] = pd.to_datetime(df_clean['order_date'])
df_clean['shipped_at'] = pd.to_datetime(df_clean['shipped_at'])
df_clean['prod_qty'] = df_clean['prod_qty'].astype(int)

# ----------------------------------
# Standardise text values
# ----------------------------------

# Remove leading/trailing spaces from order_id
df_clean['order_id'] = df_clean['order_id'].astype(str).str.strip()

# Convert prod_sku to string first, then remove spaces
df_clean['prod_sku'] = df_clean['prod_sku'].astype(str).str.strip()

# ----------------------------------
# Verify the cleaned dataset
# ----------------------------------
print("Dataset Shape After Cleaning:", df_clean.shape)

print("\nMissing Values After Cleaning:")
print(df_clean.isnull().sum())

print("\nDuplicate Records After Cleaning:")
print(df_clean.duplicated().sum())

print("\nData Types:")
print(df_clean.dtypes)

print("\nFirst Five Records After Cleaning:")
display(df_clean.head())
```

```python
# ============================================
# Date and Time Conversion
# ============================================

# Convert order_date to datetime
df_clean['order_date'] = pd.to_datetime(
    df_clean['order_date'],
    errors='coerce'
)

# Convert shipped_at to datetime
df_clean['shipped_at'] = pd.to_datetime(
    df_clean['shipped_at'],
    errors='coerce'
)

# Check for invalid dates (NaT values)
print("Missing/Invalid Dates:")
print(df_clean[['order_date', 'shipped_at']].isnull().sum())

# Fill missing shipped_at dates with the median shipment date
median_shipped_date = df_clean['shipped_at'].median()
df_clean['shipped_at'] = df_clean['shipped_at'].fillna(median_shipped_date)

# Verify that all missing dates have been handled
print("\nMissing Dates After Handling:")
print(df_clean[['order_date', 'shipped_at']].isnull().sum())

# Display data types
print("\nData Types:")
print(df_clean[['order_date', 'shipped_at']].dtypes)

# Display the first five records
print("\nFirst Five Records:")
display(df_clean[['order_date', 'shipped_at']].head())
```

```python
# ============================================
# Feature Engineering
# ============================================

# Create a copy of the cleaned dataset
df_features = df_clean.copy()

# ----------------------------------
# Order Year
# ----------------------------------
df_features['order_year'] = df_features['order_date'].dt.year

# ----------------------------------
# Order Month
# ----------------------------------
df_features['order_month'] = df_features['order_date'].dt.month

# ----------------------------------
# Order Day
# ----------------------------------
df_features['order_day'] = df_features['order_date'].dt.day

# ----------------------------------
# Day of Week
# ----------------------------------
df_features['day_of_week'] = df_features['order_date'].dt.day_name()

# ----------------------------------
# Order Hour
# ----------------------------------
df_features['order_hour'] = df_features['order_date'].dt.hour

# ----------------------------------
# Shipping Delay (Hours)
# ----------------------------------
df_features['shipping_delay_hours'] = (
    (df_features['shipped_at'] - df_features['order_date'])
    .dt.total_seconds() / 3600
)

# ----------------------------------
# Shipping Delay (Days)
# ----------------------------------
df_features['shipping_delay_days'] = (
    (df_features['shipped_at'] - df_features['order_date'])
    .dt.total_seconds() / (24 * 3600)
)

# ----------------------------------
# Shipment Status
# ----------------------------------
df_features['shipment_status'] = np.where(
    df_features['shipped_at'].notna(),
    'Shipped',
    'Not Shipped'
)

# ----------------------------------
# Display the new features
# ----------------------------------
print("Dataset Shape After Feature Engineering:", df_features.shape)

print("\nNew Features Added:")
display(df_features[['order_year',
                     'order_month',
                     'order_day',
                     'day_of_week',
                     'order_hour',
                     'shipping_delay_hours',
                     'shipping_delay_days',
                     'shipment_status']].head())

print("\nFirst Five Records:")
display(df_features.head())
```

```python
# ============================================
# Data Validation
# ============================================

# Verify data types
print("Data Types:")
print(df_features.dtypes)

# ----------------------------------
# Confirm missing values
# ----------------------------------
print("\nMissing Values:")
print(df_features.isnull().sum())

# ----------------------------------
# Check invalid product quantities
# ----------------------------------
print("\nRecords with Invalid Product Quantity (<= 0):")
display(df_features[df_features['prod_qty'] <= 0])

# ----------------------------------
# Check negative shipping delays
# ----------------------------------
print("\nRecords with Negative Shipping Delay:")
display(df_features[df_features['shipping_delay_days'] < 0].head())

print("\nTotal Negative Shipping Delays:",
      (df_features['shipping_delay_days'] < 0).sum())

# ----------------------------------
# Verify engineered features
# ----------------------------------
print("\nSummary of Engineered Features:")
display(df_features[['order_year',
                     'order_month',
                     'order_day',
                     'order_hour',
                     'shipping_delay_days']].describe())

# ----------------------------------
# Display final dataset information
# ----------------------------------
print("\nFinal Dataset Shape:", df_features.shape)

print("\nFirst Five Records:")
display(df_features.head())
```
